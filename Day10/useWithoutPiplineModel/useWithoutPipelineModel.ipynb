```python
import pickle
import numpy as np
```

```python
clf = pickle.load(open("clf.pkl",'rb'))
ohe = pickle.load(open("ohe.pkl",'rb'))
ohe_embarked= pickle.load(open("ohe_embarked.pkl",'rb'))

```

```python
test_input = np.array(
                [1, 'male', 65.0, 0, 0, 26.55, 'S'],
                dtype=object
            ).reshape(1,7)

test_input
```

```python
test_sex = ohe.transform(test_input[:,1].reshape(1,1))

test_sex
```

```python
test_embarked = ohe_embarked.transform(test_input[:,-1].reshape(1,1))
test_embarked
```

```python
test_data = np.concatenate((test_input[:,[0,2,4,5]],test_input[:,[3]],test_sex,test_embarked),axis=1)
test_data
```
