```python
import http.client

conn = http.client.HTTPSConnection("indeed-scraper-api.p.rapidapi.com")

payload = "{\"scraper\":{\"maxRows\":15,\"query\":\"Developer\",\"location\":\"San Francisco\",\"jobType\":\"fulltime\",\"radius\":\"50\",\"sort\":\"relevance\",\"fromDays\":\"7\",\"country\":\"us\"}}"

headers = {
    'x-rapidapi-key': "b64c480237msh8ab652c8f37e954p1fff34jsn0e857e58d76f",
    'x-rapidapi-host': "indeed-scraper-api.p.rapidapi.com",
    'Content-Type': "application/json"
}

conn.request("POST", "/api/job", payload, headers)

res = conn.getresponse()
data = res.read()

print(data.decode("utf-8"))

final_data = data.decode("utf-8")
```

```python
import json
response = json.loads(final_data)
print(json.dumps(response, indent=2)[:5000])
```
