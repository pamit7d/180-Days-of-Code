## Exploratory Data Analysis (EDA)

### Univariate Analysis

#### 1. Categorical Data
- Only two option in General: 1. Counterplot(Bar)  2. PieChart (extra info. of percentage intution)


- Survival Count Graph

```python
import seaborn as sns

sns.countplot(data=df, x='Survived')
```

- Out of All passangers only 38.4% survived

```python
df['Survived'].value_counts().plot(kind='pie', autopct='%1.1f%%')
print(df['Sex'].value_counts())
```

- 55.1% was in lowest class (3rd class)
- 20.7% were in middle class (2nd class)
- 24.2% were in top class (1st class)

```python
df['Pclass'].value_counts().plot(kind='pie', autopct='%1.1f%%')
```

- Male female Percentage ratio
- There were 64.76% male & 35.24% female

```python
df['Sex'].value_counts().plot(kind='pie', autopct='%.2f')
```

- Survival count of men with respect to, in Passanger class they were travelling.
- Out of 577 male, only 109 Survived.

```python
print(len(df[df['Sex'] == 'male']))
data = df[(df['Sex'] == 'male') & (df['Survived'] == 1)]
print(len(data))

sns.countplot(data=data, x='Pclass')
```

- Survival count of Women with respect to, in Passanger class they were travelling.
- Out of 314 female, 233 Survived.

```python
df['Sex'].groupby(df['Pclass']).value_counts().plot(kind='bar')
```

```python
print(len(df[df['Sex'] == 'female']))
fdata = df[(df['Sex'] == 'female') & (df['Survived'] == 1)]
print(len(fdata))

sns.countplot(data=fdata, x='Pclass')
```

- DataSet

```python
tips = sns.load_dataset('tips')

flights = sns.load_dataset('flights')

iris = sns.load_dataset('iris')
```

```python
flights.head()
```

- ScatterPlot (Numerical to Numerical)


```python
tips
```

- Here, We are doing Multivariate analysis and seeing a linear relationship between the columns in the data.

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', size='size', style='smoker')
```

- LinePlot (Numerical to Numerical)

```python
flights.groupby('year')['passengers'].sum().plot(kind='line', marker='o', title='Total Passengers per Year')
```

- Bar Plot (Numerical to Categorical)


```python
df.head()
```

- Avg. Fare of the Survived and Non-Survived Class wise Graph.

```python
sns.barplot(data=df, x='Pclass', y='Fare', hue='Survived')
```

- Avg. Age of the Survived and Non-Survived Class wise Graph.

```python
sns.barplot(data=df, x='Pclass', y='Age', hue='Survived')
```

- Box Plot (Numerical to Categorical)

```python
sns.boxenplot(data=tips, x='day', y='total_bill', hue='sex')
```

- kdeplot (Numerical to Categorical)

- This tells:
    - Take all passengers, group them by whether they survived or not, calculate the age distribution for each group, draw a smooth density curve for each group, and fill the area under the curves so they can be compared easily."

```python
sns.kdeplot(
    data=df,
    x='Age',
    hue='Survived',
    fill=True
)
```

- HeatMap (Categorical to Categorical)


```python
df.head()
```

```python
sns.heatmap(pd.crosstab(df['Pclass'], df['Survived']), annot=True, fmt='d', cmap='YlGnBu')
```

- ClusterMap (Categorical to Categorical)


```python
sns.clustermap(pd.crosstab(df['Parch'], df['Survived']), annot=True, fmt='d', cmap='YlGnBu')
```
